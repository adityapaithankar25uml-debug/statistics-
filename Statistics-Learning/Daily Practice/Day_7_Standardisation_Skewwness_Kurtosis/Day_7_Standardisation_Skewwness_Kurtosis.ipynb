Skewnness - How the data points are destributed across the graph or Mean/median

-When Skew is strong Prefer the Median Over mean

- Right Skew -
       Mean > Median
       Positive 
       tail points are right
       ex- Income , house price
- Left Skew -
        Mean < Median
        negative
        Tail points are Left 
        ex- Exam Marks
- Symmentric skew-
        Mean ~ Median
        balanced
        ex-height

```python
from scipy.stats import skew
import numpy as np 

income= np.array([22,25,27,28,30,31,33,35,40,95])

print(f"Mean = {income.mean():.1f}")
print(f"Meadin = {np.median(income):.1f}")
print(f"Skew = {skew(income):.2f}")
```
