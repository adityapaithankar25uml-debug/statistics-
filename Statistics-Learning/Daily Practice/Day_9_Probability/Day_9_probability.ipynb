 Formula : P(A|B) = P(B|A) * P(A)  / P(B)

 P(B) = Never be zero

 -  prior P(A) :  Beleive before seeing the data 
 - Likelihood P(B|A)  : How will the cause explins the evidence 
 - Posterior P(A|B) :  The Updated belief the data 
  - evidence P(B)  : Normalise the that makes the posterior  a valid probability 
  

```python
def bayes(prior,likelihood,false_alarm):
    evidence = likelihood * prior + false_alarm * (1-prior)
    return  likelihood *  prior /  evidence 

    post = bayes(prior = 0.01,likelihood=0.99,false_alarm=0.05)
    print(f"P(disease|position):{post:.4f}")
```
